#  <a name="5">5. Проверка гипотез </a>

* H0: Цены красного вина в диапазоне от 10 до 100 долалров за бутылку из США и Франции одинаковые.
* H1:Цены красного вина в диапазоне от 10 до 100 долалров за бутылку из США и Франции различны.
* Пороговое значение *alpha* - 0.05.

```python
red_wine = wine.query('color == "red" & price >= 10 & price <= 100')  
red_wine.shape
```

```python
france_prices = red_wine[red_wine['country'] =='France']
us_prices = red_wine[red_wine['country'] =='US']
```

```python
france_prices.shape
```

```python
us_prices.shape
```

Описательная статистика цен на красное вино в США и Франции.

```python
france_prices['price'].describe().round(2)
```

```python
us_prices['price'].describe().round(2)
```

Большинство значений у нас ниже среднего и по ценам во Франции и в США, предполагаем смещение вправо. Для визуализации построим график распределения.

```python
plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
sns.histplot(france_prices['price'].head(500), kde=True, bins=30)
plt.title('Распределение цен на красное вино в Европе')
plt.xlabel('Цена')

plt.subplot(1, 2, 2)
sns.histplot(us_prices['price'].head(500), kde=True, bins=30)
plt.title('Распределение цен на красное вино в Северной Америке')
plt.xlabel('Цена')
plt.figtext(0.4, 0, "Рисунок 18. - Распределение цена на красное вино")
plt.tight_layout()
plt.show
```

Данные смещены с отклонением вправо (положительное распределение).

Для смещенных данных лучше использовать непарметрические методы. Например, U-критерий Манна-Уитни или тест Крускала-Уоллиса.

Для теста Манна-Уитни желательно чтобы объем выборок был не более 30 наблюдений. Сформулируем случайную выборку с помощью метода .sample(n=...) 

```python
np.random.seed(42)
sample_france_prices = france_prices.sample(n=30, random_state=42)
sample_us_prices = us_prices.sample(n=30, random_state=42)
print(f"Размер выборки красного вина из Франции:{len(sample_france_prices)}")
print(f"Размер выборки красного вина из США:{len(sample_us_prices)}")
```

Выведем статистику по новым выборкам:

```python
sample_france_prices.describe()
```

```python
sample_us_prices.describe()
```

Применим к новым выборкам тест Манна-Уитни

```python
u_stat, p_value = st.mannwhitneyu(
    sample_france_prices['price'],
    sample_us_prices['price'],
    alternative='two-sided'
)
print(f"U-статистика: {u_stat}, p-значение: {p_value:.4f}")
        
```

p < 0.05 следовательно нулевую гипотезу **отвергаем**. Есть статистически значимые различия между распределениями двух выборок. Цены красного вина из Франции и США **различаются**.

Проверим гипотезу с помощью  t-критерия Уэлча, т.к. у нас размеры выборок не равны и дисперсии в группах существенно различаются. 

```python
st.ttest_ind(france_prices['price'], us_prices['price'], equal_var=False)
```

pvalue < 0,05 следовательно нулевая гипотеза **отвергается** и принимается гипотеза, что цены красного вина в  диапазоне от 10 до 100 долларов из Франции и США **различны**.


Таким образом, проденное исследование с помощью двух критериев тест Манна-Уитни и t-критерия Уэлча подтвердило, что цены красного вина в диапазоне от 10 до 100 долларов из Франции и США различаются.
